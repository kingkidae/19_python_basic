```python
#while 반복문 사용하기
'''
초기식
while 조건식:
    반복할 코드
    변화식
'''
i = 0
while i < 100:
    print("hello world", i)
    i += 1 # => i+i = 1
```

```python
# 초기값을 1부터 시작하기
i = 1
while i <= 10:
    print("hello.", i)
    i += 1
```

```python
# 초기값을 감소시키기
i = 10
while i > 0:
    print("hi.", i)
    i -= 1
```

```python
# 입력한 횟수대로 반복하기
count = int(input("repeat"))
i = 0
while i < count:
    print("hi", i)
    i += 1
```

```python
# 반복 횟수가 정해지지 않는 경우
import random

random.random() #랜덤이라는 모듈 안의 랜덤이라는 함수를 실행하는 명령문. 실행하면 0에서 1사이의 수를 아무거나 준다.
```

```python
random.randint(1, 6) # 설정한 범위 내에서 랜덤한 숫자를 반환함. 끝에 들어간 숫자도 포함함. 랜덤 병닉값 제대로...
```

```python
import random

i = 0
while i != 3:
    i = random.randint(1, 6)
    print(i)
```
