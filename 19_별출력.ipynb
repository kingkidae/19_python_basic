```python
# 중첩 루프 사용하기
for i in range(5):
    for j in range(5):
        print("j:", j, sep = "", end = " ")
    print("i: ", i, "\\n", sep = "") # \ \n 역슬래시를 문자모양 그대로 쓰려고 역슬래시가 두개가 됨.
```

```python
# 사각형으로 별 출력하기
for i in range(5): # i는 변수의 역할은 안하고 있음. 0-4, 0-4들어가 있어서 25번 반복될거임.
    for j in range(5):
        print("*", end = "") # 별찍고 엔드파라미터를 없애버리니까 5개의 별을 찍고 끝나는데
    print() #프린트까지 마치고 나면 다시 for i 코드로 이동해서 0-4까지 반복함.
```

```python
# 사각형 모양 바꾸기
for i in range(3): #세로
    for j in range(7): #가로
        print("*", end = "")
    print() # i가 바뀔 때 출력을 하라는 뜻
```

```python
# 계단식으로 별 출력하기
for i in range(5):
    for j in range (5):
        if j <= i:
            print("*", end = "")
    print()
```
