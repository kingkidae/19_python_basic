```python
# for와 range 사용하기
'''
for 변수 in range(반복할 횟수):
    반복할 코드
'''
for i in range(100):
    print("hello world!")
```

```python
# 반복문에서 변수의 변화 알아보기
for i in range(10):
    print("hello, world", i)
```

```python
# 시작하는 숫자와 끝나는 숫자 지정하기
for a in range(5, 12):
    print("hell", a)
```

```python
# 증가폭 사용하기
for b in range(0, 10, 2):
    print("hell", b)
```

```python
# 숫자 감소시키기
for c in range(10, 0, -1):
    print("hei", c)
```

```python
a = [1, 3, 5, 7, 9] #그냥 리스트 내용을 뒤집는 것임. 오름/내림 차순이 아님.
list(reversed(a))
```

```python
for i in reversed(range(10)):
    print(i)
```
