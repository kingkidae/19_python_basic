```python
# 음료수 자판기 만들기
button = int(input())
# elif 대신에 if를 써도 해당 문제에서 답은 같게 나올 수 있겠지만 사실 if로 쓰면 여러번 실행하는 것이고, elif를 사용한다면 한번 실행하는 것임. 
if button == 1:
    print("coke")
elif button == 2:
        print("sprite")
elif button == 3:
        print("fanta")
else:
    print("wrong menu")
```

```python
age = int(input())
card = 9000
if 7 <= age <= 12:
    print(card - 650, "원")
elif 13 <= age <= 18:
    print(card - 1050, "원")
elif 19 <= age:
    print(card - 1250, "원")
else:
    print("wrong age")
```

```python
age = int(input())
card = 9000
if 7 <= age <= 12:
    card = card - 650
elif 13 <= age <= 18:
    card = card - 1050
elif 19 <= age:
    card = card - 1250
else:
    print("wrong age")
print(card, "원")
```
