```python
s = "the grown-ups' responce, thos time, was so advise me to lay aside"
s = list(s.split())
s.count("the")
```

```python
s = list(map(int, input().split(";")))
s.sort()
s.reverse()
s
```

```python
for i in s:
    print(f"{i:8}") # i뒤에 콜론을 적고 숫자를 적으면 8자리를 확보하고 나머지에는 공백
```

```python
for i in s:
    print(f"{i:0.4f}") # i뒤에 콜론을 적고 소수를 적고, f적으면 소수점 네자리까지
```

```python
for i in s:
    print(f"{i:10,.4f}") # i뒤에 콜론을 적고 숫자를 적고 소수를 적고, 공백 +f적으면 소수점 네자리까지 숫자 뒤에 , 적으면 천자리 구분 가능
```
