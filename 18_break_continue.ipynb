```python
# while반복문에서 break반복문 끝내기

i = 0
while True:
    print(i)
    i += 1
    if i == 5:
        break
```

```python
# for반복문에서 break 반복문 끝내기

for i in range(100):
    print(i)
    if i == 10:
        break
```

```python
#continue로 코드 실행 건너뛰기
for i in range(10):
    if i % 2 == 0:
        continue # 밑에 명령문을 무시하고 위의 명령문(코드)으로 이동하게 함.
    print(i)
```

```python
i = 0
while i < 5:
    i += 1
    if i % 2 == 0: # i를 2로 나눴을 때 나머지가 0 => 짝수
        continue 
    print(i)
```

```python
# 입력한 횟수대로 반복하기
count = int(input("repeat"))

i = 0
while True: # 이대로 두면 무한루프임.
    print(i)
    i += 1 # i = i + 1
    if i == count:
        break
```

```python
# 입력한 숫자까지 홀수 출력하기
count = int(input())
for i in range(count + 1):
    if i % 2 == 0:
        continue
    print(i)
```

```python
i = 0
while True:
    if i % 10 != 3:
        i += 1
        continue
    if i > 73:
        break
    print(i, end = ' ')
    i += 1
```

```python
start, stop = map(int, input().spllit())

i = start

while True : 
    if i % 10 == 3:
        i += 1
        continue
    if i > stop:
        break
    print(i, end = " ")
    i += 1
```
