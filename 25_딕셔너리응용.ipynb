```python
x = {
    "a" : 10, # 키와 밸류 구분은 콜론으로 되어 있음.
    "b" : 20,
    "c" : 30,
    "d" : 40
}
```

```python
x.get("a") # x["a"]와 동일함.
x.get("z") # x["a"]와 동일함.get으로 하면 z가 없음에도 에러가 안남.
```

```python
x["z"] 동일한 방법으로 z를 찾아보면 에러 처리가 남.
```

```python
#딕셔너리에서 키:값 쌍을 모두 가져오기
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
x.items()
x.keys()
x.values()
```

```python
# 반복문으로 딕셔너리의 키:값 모두 출력하기

for i in x:
    print(i) #i는 임의의 변수, x는 딕셔너리 데이터
```

```python
for i in x.items():
    print(i)
```

```python
for key, value in x.items():
    print(key, value)
```
