```python
#불과 비교 연산자 사용하기
print(True) #True/False의 앞글자는 무조건 대문자로 적어야 함.
print(type(True))
print(False)
print(type(False))
```

```python
#비교연산자의 판단 결과
print(3 > 1)
print(3 < 1)
```

```python
# 숫자가 같은지 다른지 비교
print(10 == 10)
print(10 != 10) # !=는 다르다는 뜻
```

```python
print(10 > 20)
print(10 < 20)
print(10 >= 10)
print(10 <= 10)
```

```python
#논리연산자 사용하기
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
# or 논리연산자
print(True or True)
print(True or False)
print(False or True)
print(False or False) #한번에 바꿔야 할 때 alt + 클릭해서 일괄 편집 가능
```

```python
not True and False or not False #우선 순위는 not이 우선된다.
```

```python
# 논리연산자와 비교연산자를 함께 사용하기
print(10 == 10 and 10 != 5)
print(10 > 5 or 10 < 3)#비교연산자가 논리연산자보다 우선 된다. 모르겠으면 그냥 괄호치셈.
print(not 10 > 3) # 10 > 3의 참 거짓을 먼저 판별 후 참이 나왔는데 not이니까 답이 false가 됨.
not (1 is 1.0) #is는 같은 주소인지 여부를 확인하는 것임. 즉 숫자는 같아 보여도 메모리 주소까지 같은건지 물어보는 것임.
```
