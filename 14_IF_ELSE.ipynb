```python
# if 조건문의 동작 방식 알아보기
if True: 
    print("true")
else:
    print("false")
if False: 
    print("true")
else:
    print("false")
if None: #None은 False와 같다.
    print("true")
else:
    print("false")
```

```python
print(bool(1)) # 숫자에서는 0을 제외하고 모두 True이다.
print(bool(100))
print(bool(0))
print(bool("python"))
print(bool("high"))
print(bool("")) # 문자열에서는 빈 문자열을 제외하고 모두 True이다. 공백도 True이다.
print(bool(1.5)) # 실수도 True다.
print(bool([1, 2]))
print(bool([])) # 리스트, 튜플 등등 빈 거 빼고 트루
```
