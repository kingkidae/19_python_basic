```python
# 딕셔너리 만들기
'''
딕셔너리 = {
    키1 : 값1, 
    키2 : 값2, ...
}

'''
lux = {
    "Health" : 490,
    "mana" : 334,
    "melee" :550,
    "armor" : 18.72
}
print(lux)
```

```python
#키 이름이 중복이 되면
lux = {
    "Health" : 490,
    "Health" : 800, # 중복되었을 때 마지막 키 밸류 요소를 읽는다.
    "mana" : 334,
    "melee" :550,
    "armor" : 18.72
}
print(lux)
```

```python
# 딕셔너리 키의 자료형
x = {
    100: "hunnit", #숫자 가능
    False: 0,#불 형태 가능
    3.5: [3.5, 3.5]#리스트 가능
}
x
```

```python
x={
    [10, 20] : 100 #뉴터블에 데이터 타입은 키로써 사용불가 리스트의 경우에는 안의 내용을 변경 및 삭제가 가능해서. 튜플은 추가 수정 삭제가 안되므로 키 값으로 가능. 밸류는 어떤 형태든 가능
}
x
```

```python
#딕셔너리의 키에 접근하고, 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux['health'] # 키값을 이용하면  요소 전체가 아닌 벨류의 값만 가지고 옴.
print(lux['health'])
print(lux['mana'])
print(lux['melee'])
print(lux['armor'])
```

```python
#딕셔너리의 키에 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}

lux['health'] = 2037
lux['mana'] = 1184

lux
```
