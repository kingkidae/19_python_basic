```python
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0: # 3, 5 공배수
        print("FIzzBuzz")
    elif i % 3 == 0: #3의 배수라는 뜻
        print("Fizz")
    elif i % 5 == 0: #5배수
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 101):
    print("fizz" * (i % 3 == 0) + "buzz" * (i % 5 == 0) or i) #false는 0이고, true는 1이다. 그러므로 피즈나 버즈에 해당이 안될 경우 문자열에 0을 곱하니까 출력이 안되는 거임. or i는 그냥 숫자를 쓰는거임. 둘다 해당될 경우에는 피즈랑 버즈가 둘다 적히는데 중간에 +가 있으니까 피즈버즈 되는거임.
```
