```python
print(1, 2, 3, sep = "//") #1과 2와 3은 인수라고 부르는데 이 인수를 구분하기 위해 콤마를 적음. sep(seperate parameter)는 구분자임
```

```python
print(4, 5, 6, sep = ", ")
print("hello", "python")
print(1920, 1080, sep = "X")
```

```python
#줄 바꿈 활용하기
```

```python
print(1, 2, 3)
```

```python
print(1, 2, 3, sep = "\n") # \n은 이스케이프 코드이고, 뉴라인이라는 뜻으로 줄바꿈하라 라는 뜻임.
```

```python
print("1\n2\n3")
```

```python
#엔드 파라메타 end 사용하기
print(1)
print(2)
print(3)
```

```python
print(1, end = " ")
print(2, end = " ")
print(3, end = " ")
#원래 end파라미터는 기본적으로 \n으로 설정되어 있음. 그래서 세로로 값이 나오는 것임. 구분자를 다른 방식으로 바꿔줄 수 있음.
```

```python
y, m, d, h, mi, s = input().split()
print(y, m, d, sep = "-", end = "T")
print(h, mi, s, sep = ":")
```
