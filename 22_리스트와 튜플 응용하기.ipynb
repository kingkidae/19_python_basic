```python
# 리스트 표현식 사용하기
'''
리스트 컴프리헨션(축약식)이라고도 부름
[식 for 변수 in 리스트]
'''

a = [i for i in range(10)]
a
```

```python
a = []
for i in range(10):
    a.append(i)
a
```

```python
c = []
for i in range(10):
    c.append(i + 5)
print(c)
```

```python
[i + 5 for i in range(10)]
```

```python
# 리스트 컴프리헨션(표현식)에서 if 조건문 사용하기
b = []
for i in range(10):
    if i % 2 == 1:
        b.append(i + 5)
b
```

```python
[i + 5 for i in range(10) if i % 2 == 1]
print(b)
```

```python
#for 반복문과 if 조건문 여러 번 사용하기
[i*j for i in range(2, 10) for j in range(1, 10)]
```

```python
c = []
for i in range(2, 10):
    for j in range(2, 10):
        c.append(i * j)
print(c)
```
