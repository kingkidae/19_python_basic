```python
#위치인수
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)
```

```python
print_numbers(10, 20, 30)
```

```python
#언패킹 사용하기

x = [10, 20, 30]
print_numbers(*x) # *x = *[10, 20, 30]
```

```python
#가변인수 함수 만들기 (많이 쓰임)
''' 
def 함수이름(*매개변수):
    코드
'''

def print_numbers(*args):
    print(args)
    for arg in args:
        print(arg) #arg는 임의로 생성한 변수 암거나 적어도 됨.
```

```python
print_numbers(10)
```

```python
def print_numbers(*args):
    print(args)
    for arg in args:
        print(arg) # 얜 튜플로 나옴
print_numbers(10, 20, 30, 40)
```

```python
def personal_info(name, age, address):
    print("name", name)
    print("age", age)
    print("add.", address)
```

```python
x = {
    "name": "HKD",
    "age" : "30",
    "address" : "yong-san"
}

personal_info(**x) #왜 별표 두개냐 -> 두번 언패킹해서
personal_info(name = 'HKD', age = 30, address = 'yong-san')
```

```python
#가변 키워드 인수 함수 만들기
'''
def 함수이름(**매개변수):
    코드
'''

def personal_info(**kwargs):
    print(kwargs) #keyword Arguments
    for kw, arg in kwargs.item():
        print(kw, arg)
    
```

```python
personal_info(name = "홍길동", age = 30)
```

```python
#매개변수의 초깃값 지정하기

def personal_info(name, age, address="비공개"):
    print("name", name)
    print("age", age)
    print("add.", address)
```
