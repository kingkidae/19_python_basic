```python
# map()
a, b = map(int, input("숫자 두 개를 입력하세요").split()) #iterable(꺼내 쓸 수 있는)
# 위에 명령문 실행하면 <map at 0x1fe5a322340> 이렇게 나옴
# map(int, ["10, 20"])과 동일하고 = 10, 20이런 값이 나옴.
print(a + b)
```
