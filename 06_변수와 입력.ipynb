```python
# 변수 여러개를 한번에 만들기
x, y, z = 10, 20, 30
# 이건 튜플 데이터 타입이다. 튜플, 딕셔너리, 리스트는 괄호로 묶여있고, 꺼내 쓸 수 있는 변수이다.

print(x)
print(y)
print(z)
```

```python
x = y = z = 10
print(x)
print(y)
print(z)
```

```python
x, y = 10, 20 #위에서 아래로 읽어서 실행하기 때문에 이런 값이 나옴
x, y = y, x
print(x)
print(y)
```
